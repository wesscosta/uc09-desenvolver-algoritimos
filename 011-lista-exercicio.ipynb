04. Um sistema armazena 10 palavras digitadas pelo usuários em uma lista. Para uma análise linguística simples, deseja-se saber quantas dessas palavras possuem
mais de cinco letras. Desenvolva um programa que percorra a lista e faça essa
contagem.

```python
lista = []

for repeticao in range(10):
    palavra_digitada = input(f"Digite {repeticao+1}ª palavra: ")
    lista.append(palavra_digitada)

print(lista)
# ['Pimenta', 'Café', 'Arroz', 'Feijão', 'Açucar', 'Carne', 'Cheiro Verde', 'Flocão', 'Tomate', 'Lasanha']

for temp in lista:
    if len(temp) > 5:
        print(f"{temp} (tem {len(temp)} letras)")
```

```python
lista = []
nova_lista = []

for repeticao in range(10):
    lista.append(input(f"Digite {repeticao+1}ª palavra: "))

for temp in lista:
    if len(temp) > 5: nova_lista.append(temp)

print(f"Lista completa: {lista}")
print(*nova_lista, sep=", ")
```

05. Uma aplicação precisa organizar números informados pelo usuário de acordo com sua paridade. O programa deve solicitar dez números inteiros, armazená-los em uma lista principal e, posteriormente, separá-los em duas novas listas uma contendo apenas números pares e outra contendo apenas números ímpares. Ao final, todas as listas devem ser exibidas.

```python
nume = []
impares = []
pares = []
for i in range(6):
    n = int(input("digite um numero"))
    nume.append(n)
    if n % 2== 0: 
        pares.append(n)
    else:
        impares.append(n)
    
print(*nume)
print(pares)
print(impares)
```

06. Um site esportivo permite que o usuário informe nomes de esportes que pratica. Os dados devem ser armazenados até que o usuário digite a palavra “fimˮ. Após o encerramento da entrada, o sistema deve mostrar quantos esportes foram cadastrados e exibi-los em ordem alfabética.

```python
lista = []
while True:
    antonio = input("informe a modalidade")
    if antonio == "fim":
        break
    lista.append(antonio)
lista.sort()
print(lista)
```

08. Um laboratório registrou oito medições numéricas. Após armazenar esses valores em uma lista, o sistema deve calcular a média aritmética e identificar
quais medições ficaram acima dessa média, exibindo apenas esses valores.

```python
listas = []
for n in range(8):
    medicao = float(input("Digite as medições:\n"))
    listas.append(medicao)
for lista in listas:
    if lista>(sum(listas)/len(listas)):
        print(f"A média é {sum(listas)/len(listas)} e essas medições estão acima da média: {lista}")
```
