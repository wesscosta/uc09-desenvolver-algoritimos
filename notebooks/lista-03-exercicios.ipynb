09. Leia do usuário e monte um dicionário compra com: produto , preco (float), qtd (int). Calcule e mostre total. Depois, se total >= 100 , mostre “frete grátis”, senão “frete normal”

```python
produto = input("Insira um produto:\t")
preco = float(input("Insira o preço:\t"))
quantidade = int(input("Insira a quantidade:\t"))

compra = {
    "Produto" : produto,
    "Preço" : preco,
    "Quantidade" : quantidade
}

total = compra["Preço"] * compra["Quantidade"]

if total >= 100:
    print("Você ganhará frete grátis!")
else:
    print("O valor da sua compra é de {total}")
```

10 . Crie um dicionário fixo estoque = {"arroz": 10, "feijao": 5, "cafe": 2}. Peça um item e uma quantidade (int).
Se existir no estoque e tiver quantidade suficiente, atualize o estoque
(subtraia). Se não existir ou não tiver, mostre mensagem.

```python
estoque = {"arroz": 10,
           "feijao": 5,
           "cafe": 2
}

item = input("digite o item:")

if item in estoque:
    quant = int(input("digite uma quantidade: "))
    if estoque[item] >= quant:
        estoque[item] -= quant
        print(estoque)
    else:
        print("quantidade insuficiente.")
else:
    print("item nao existe no estoque")
```

16. Cadastro até “sair” (lista de dicionários)
Crie cadastros = [] .
Enquanto o usuário não digitar sair :
ler nome
ler idade (int)
ler cidade
salvar em um dicionário e adicionar na lista
No final, mostre:
total de pessoas cadastradas
média das idades

```python
cadastro = []

while True:
    nome = input("Insira seu nome:\t").title()
    if nome == "Sair": break
    idade = int(input("Insira sua idade:\t"))
    cidade = input("Insira sua cidade:\t").title()
    if cidade == "Sair": break
    cadastro.append({
        "nome" : nome,
        "idade" : idade,
        "cidade" : cidade
    })

print(len(cadastro))

total = 0

for i in cadastro:
    total += i["idade"]

media = total / len(cadastro)
print(media)
```
