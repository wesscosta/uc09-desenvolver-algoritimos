02. Crie um programa que solicite ao usuário dois números e exiba o resultado da soma, subtração, multiplicação e divisão desses números.



```python
num1 = float(input("digite o primeiro numero: "))
num2 = float(input("digite o segundo numero: "))

soma = num1 + num2
subtracao = num1 - num2 
multiplicacao = num1 * num2 
divisao = num1 / num2


print("#"*10,"__ Resultado__","#"*10)
print(f"soma: {soma}")
print(f"subtração: {subtracao}")
print(f"multiplicação: {multiplicacao}")
print(f"divisão: {divisao}")
print("#"*50)
```
