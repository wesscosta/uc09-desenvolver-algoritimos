Questão 2 - Durante o desenvolvimento de um sistema, o programador precisa compreender a diferença entre números inteiros e números reais. Crie um programa que permita ao usuário informar um número inteiro e, em seguida, um número decimal. Após a leitura, o programa deve exibir os valores informados e também indicar o tipo de dado de cada um. 

```python
numero_inteiro = int(input("digite um numero inteiro"))
numero_decimal = float(input("digite um numero decimal"))

print(f"O numero {numero_inteiro} é do tipo {type(numero_inteiro).__name__}")
print(f"O numero {numero_decimal} é do tipo {type(numero_decimal)}")
```

6. Um sistema de apoio ao aprendizado de matemática precisa realizar operações 
básicas entre dois números informados pelo usuário. O programa deve ler dois 
valores numéricos e exibir o resultado da soma, subtração, multiplicação e divisão entre eles.

```python
num1 = float(input("digite o 1ª valor"))
num2 = float(input("digite o 2º valor"))

print(f" SOMA: {num1} + {num2} = {num1+num2}")
print(f" SUBTRAÇÃO: {num1} - {num2} = {num1-num2}")
print(f" MULTIPLICAÇÃO: {num1} * {num2} = {num1*num2}")
print(f" DIVISÃO: {num1} / {num2} = {num1/num2}")
```

7. Um professor deseja automatizar o cálculo da média de seus alunos. Crie um 
programa que permita informar três notas e, ao final, calcule e exiba a média 
aritmética dessas notas. 


```python
# Entrada de notas
nota1= float(input("informe a primeira nota:"))
nota2= float(input("informe a segunda nota:"))
nota3= float(input("informe a terceira nota:"))
# calculo das notas
media = (nota1 + nota2 + nota3) / 3
print("---Resultado---")
print(f"Média aritmética: {media}")
```

09. Uma empresa deseja calcular o salário líquido de seus funcionários. Para isso, o sistema deve permitir informar o salário bruto e o percentual de desconto aplicado. O programa deve calcular o valor do desconto e exibir o salário final após o desconto.

```python
salario_bruto = float(input("Digite o salario bruto"))
desconto = float(input("Digite o desconto"))
salario_liquido = salario_bruto * (2-(desconto/100))
print(f"salario liquido: {salario_liquido}")
```
