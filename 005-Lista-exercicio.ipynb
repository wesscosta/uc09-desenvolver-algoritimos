14. Um sistema de segurança controla o acesso de usuários com base em duas condições. O programa deve solicitar a idade do usuário e se ele possui autorização. O acesso só deve ser permitido caso a idade seja maior ou igual a 18 anos e a autorização seja verdadeira. O sistema deve informar se o acesso foi liberado ou negado. 

```python
#Leitura de dados
idade=int(input("Digite sua idade"))
autorizacao=input("Você possui autorização")
#conversão
tem_autorizacao = autorizacao.lower() == "s"

#resultados
if idade>= 18 and tem_autorizacao:
    print("acesso liberado")
else:
    print("acesso negado")
```
